# EDA - confessionals and the edit

Requires `uv run snuffml build` first.

```python
import pandas as pd
import matplotlib.pyplot as plt
from snuffml.features import build

df = build.load_features()
alive = df[df['in_game']]
df.shape, sorted(df['season'].unique())[-5:]
```

```python
# winner vs non-winner cumulative conf share at the finale, split by era
finale = alive[alive['episode'] == alive.groupby('season')['episode'].transform('max')]
ax = finale.boxplot(column='conf_share_cum', by=['era', 'is_winner'], figsize=(9, 5), rot=30)
ax.set_title('Final-episode cumulative confessional share'); plt.suptitle('');
```

```python
# the famous new-era flattening: winner share by season
w = finale[finale['is_winner']].sort_values('season')
plt.figure(figsize=(10, 4))
plt.bar(w['season'], w['conf_share_cum'], color=w['era'].map({'old': '#888', 'middle': '#4a7', 'new': '#c55'}))
plt.axvline(40.5, ls='--', c='k'); plt.xlabel('season'); plt.ylabel('winner conf share');
```

```python
# how early does the winner's edit separate? mean conf_z_cum by season progress
alive2 = alive.assign(frac=(alive['episode'] / alive.groupby('season')['episode'].transform('max') * 10).round() / 10)
alive2.groupby(['frac', 'is_winner'])['conf_z_cum'].mean().unstack().plot(figsize=(9, 5), marker='o');
```
